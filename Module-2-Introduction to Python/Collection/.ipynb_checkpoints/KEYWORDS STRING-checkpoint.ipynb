```python
'''WHAT IS STRING ?
STRING IS A SET OF CHARACTERS ENCLOSED IN '' OR "" OR ''' ''' OR """ """.
STRING LITRALS : HOW TO DEFINE IT .EXAMPLES-'' OR "" OR ''' ''' OR """ """.
STRING IS IMMUTABLE-YOU CANNOT CHANGE/DELETE/ADD/REMOVE. FROM IT ONCE GIVEN OR ASSIGNED TO VARIABLE.
```

```python
abc="i am devanshu"
name='i am kd'
b="""hello ji"""
c='''devanshu is me'''
print(abc+"\n"+name+"\n"+b+"\n"+c)  # + meansstring concetination.
```

```python
#ACCESSING A SPECIFIC CHARACTER IN STRING:-
name="i am devanshu"
print(name[0])
print(name[3])
```

```python
#ACCESSING A SPECIFIC CHARACTER FROM END:-
name="i am devanshu"
print(name[-1])
print(name[-4])
```

```python
#SLICING IN STRING:-
name="i am devanshu"
print(name[1:6])    # ALSO COUNT SPACE.
```

```python
#SLICING REVERSE IN STRING:-
name="i am devanshu"
print(name[-6:-1])    # ALSO COUNT SPACE.
```

```python
#REVERCE THE STRING:-
#SLICING IN STRING:-
name="i am devanshu"
print(name[::-1])    # ALWAYS REVERSE THE FULL STRING.
```

```python
#STRING ITERATION
# NUMBER OF ITERATION REPEATED AGAIN AND AGAIN LOOP
name="hi my name is devanshu"
for i in name:   # loop always take 1 word/1 assigned number.
    print(i)
```

```python
#STRING IS A IMMUTABLE BUT THEN TOO CAN BE CHANGED.
name="hi i am devanshu"
result=name[0:8]+"DEVANSHU"    #[0:8 takes from hi i am ] THIS IS HOW STRING CAN BE CHANGED.
print(result)
```

```python
#DELETE THE STRING
name="hi i am devanshu"
del name      # del is used to delete the string assigned to variable.
print(name)
```

```python
# REPLACE THE STRING :-
name1="hi i am devanshu"
name2=name1.replace("devanshu","rutvi")
print(name1)
print(name2)
```

```python
#UPPER AND LOWER() USED IN STRING:-
name1="hi i am devanshu"
print(name1.upper())
print(name1.lower())
```

```python
#LENGTH USED IN STRING:-
name="i am devanshu"
print(len(name))
```

```python
#CONCATINATION IN STRING:-
name1="hi devanshu"
name2="i am KD"
print(name1+" "+name2)  # this is concatination
```

```python
#CONCATINATION IN STRING:-
name1="DEVANSHU"
print((name1+"\n")*10)
```

```python
# USE OF F-STRING IS FORMATING INSIDE THE PLACEHOLDER()
# EMBEDED A STRING USINH F-STRING F{}
# EXAMPLE:-
name="devanshu"
age="29"
print(f"name: {name},age: {age}")  #in placeholder={} print=variable.
```
