```python
''' WHAT IS LIST ?
--LIST IS A SET OF ORDER THAT CAN BE STORED INSIDE OF[](SQUARE BRACKETS).
--LIST PROVIDE MUTABLE(CAN BE CHANGED) OBJECTS.
--LIST IS USED TO STORE MULTIPLE DATA IN SINGLE VARIABLE.
```

```python
#EXAMPLES:-
employee=["devanshu","brijesh","rajesh"]
print(employee)
```

```python
#EXAMPLES:-
employee=["devanshu","brijesh","rajesh"]
print(employee[0])
```

```python
#ACCESSING THE LIST:-
employee=["devanshu","brijesh","rajesh"]
print(employee[0:3])
```

```python
name=[1,2,7,8,"devanshu","amit","rutvi"]
print(name)
```

```python
#HOW TO DEFINE LIST IN TUPLE():-
employee=list[(1,2,3,4,5,7,8,9,0,"devanshu","amit")]    # here in () parenthesis. it is always tuple.
print(employee)
```

```python
#ACCESING THE LIST.
employee=[1,2,3,4,5,7,8,9,0,"devanshu","amit"]    # here in () parenthesis. it is always tuple.
print(employee[-5:-1])
```

```python
#APPEND IS USED TO ADD AT LAST POSITION:-
employee=[1,2,3,4,5]
employee.append(6)
employee.append(7)
employee.append(8)
employee.append(9)
print(employee)
```

```python
#INSERT USED TO ADD VALUE IN ANY POSITION:-
a=[1,2,3,4,5]
a.insert(3,9)     #(position,value)
print(a)
```

```python
#EXTAND IS USED TO ADD MORE THAN ONE VALUES.BUT IN LIST[] ONLY.
a=[1,2,3,4,5]
a.extend([6,7,8,9,10])  # extend provide more values but bust be in list[].
print(a)
```

```python
#CLEAR IS USED TO CLEAR ENTIRE LIST.SINGLE VALUE CANNOT BE CLEAR.
a=[1,2,3,4,5]
a.clear()
print(a)
```

```python
# REMOVE IS USED TO REMOVE SINGLE VALUE:-
a=[1,2,3,4,5]
a.remove(5)
print(a)
```

```python
#UPDATE IS USED TO UPDATE THE VALUES IN LIST:-
a=[1,2,3,4,5]
a[2]=10        #a[position]=value
print(a)
```

```python
a=[1,2,3,4,5]
a[0:2]=-1,-2,-3        #a[start position:end position]=values
print(a)
```

```python
employee=["amit","rajesh","abhishek","devanshu","vijay","akshat","ruhi","rutvi"]
emp=str(input("ENTER THE EMPLOYEE NAME :")).lower()
if emp in employee:
    print("HE IS EMPLOYEE")
else:
    print("NOT AN EMPLOYEE")
```

```python
#pop is used to delete specific value in list.
a=[10,20,30]
print(a.pop(2))     # (a.pop(index))
```

```python
#del is used to delete the perticular value
a=[10,20,30]
del a[2]               # del a[index]
print(a)
```
