```python
a = [1, 2, 3]
b = a      # b points to the same object as a
c = [1, 2, 3]  # c is a new list, even though it has the same contents

# a==c is true.
# but a is c is false.
# == → compares values.
# is → compares identity (memory location).
print("a is b:", a is b)       # True, same object
print("a is c:", a is c)       # False, different objects

# Using 'is not'
print("a is not b:", a is not b)   # False b=a
print("a is not c:", a is not c)   # True
```
