```python
def emp(lst):   #lst is just taken as variable to store list in it.
    employee=str(input("ENTER THE EMPLOYEE NAME :"))
    if employee in lst:
        print("HE IS EMPLOYEE")
    else:
        print("HE IS NOT AN EMPLOYEE")
emp(["devanshu","amit","rutvi","sonu","krishna","rahul","kishan","rajiv","shrinil","mohit","rajesh","mitesh"])   #this all values we assign to lst.
```

```python
#FOR FACTORIAL OF N NUMBERS:-
def a():
    n=int(input("ENTER THE Nth NUMBER :"))
    fact=1
    for i in range(1,n+1):
        fact*=i
        print(fact)
a()
```
