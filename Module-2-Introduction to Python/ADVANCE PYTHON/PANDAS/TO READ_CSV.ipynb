```python
# USE OF TO_STRING:-
import pandas as pd
a=pd.read_csv("employee_data_100_rows.csv")    #THIS FILE HAS 100 DATA IN IT.
print(a.to_string())                           # .TO_STRING-->RETURNS ALL DATA
```

```python
#TO INCREASE MAXIMUM DATA FROM CSV:-
import pandas as pd
pd.options.display.max_rows=9999       # YOU HAVE TO GIVE 9999 SO THAT IT WILL TAKE ALL OR ELSE BY TAKING 99 IT WILL BE SIMPLY LIKE FIRST 5 AND LAST 5.
res=pd.read_csv('employee_data_300_rows.csv')
print(res)
```

```python
#.HEAD:-(UP)
import pandas as pd
res=pd.read_csv('employee_data_300_rows.csv')
print(res.head(10))          # HEAD BY DEFAULT--5 AND GIVE VALUE TO SHOW TILL DATA.
```

```python
#.TAIL:-(DOWN)
import pandas as pd
res=pd.read_csv('employee_data_300_rows.csv')
print(res.tail(10))                   # SHOWS LAST 5 DATA BY DEFAULT OR YOU CAN GIVE AS PER CHOICE.
```
