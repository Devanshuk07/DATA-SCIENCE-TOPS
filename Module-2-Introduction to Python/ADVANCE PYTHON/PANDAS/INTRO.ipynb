```python
import pandas as pd
a=pd.DataFrame({                                   #DATAFRAME D AND F MUST BE CAPITAL.
    'name':['devanshu','ravi','raj','jay'],        # ALL ARRAYS MUST BE OF SAME LENGTH.
    'salary':[20000,50000,90000,8000]
})
print(a)
```

```python
#JUST TAKING SINGLE SINGLE COLUMNS:-
import pandas as pd
a=pd.DataFrame({                                   #DATAFRAME D AND F MUST BE CAPITAL.
    'name':['devanshu','ravi','raj','jay'],        # ALL ARRAYS MUST BE OF SAME LENGTH.
    'age':[20,24,30,35],
    'salary':[20000,50000,90000,8000]
})
print(a[['name','salary']])                     # ONLY TAKING THE SELECTED COLUMNS.
print(a) 
```

```python
#COLUMNS SELECTION:-
import pandas as pd
a=pd.DataFrame({                                   #DATAFRAME D AND F MUST BE CAPITAL.
    'name':['devanshu','ravi','raj','jay'],        # ALL ARRAYS MUST BE OF SAME LENGTH.
    'age':[20,24,30,35],
    'salary':[20000,50000,90000,8000]
})
b=a.loc[a['age']<30,['name','salary','age']]     # .loc is used to select rows and columns by LABELS and CONDITIONS.
print(b)
```
