```python
'''JSON STANDS FOR JAVASCRIPT OBJECTS NOTATION.
A PANDAS USED TO READ A JSON FILE CONTENT IN PYTHON.
A JSON CONTAIN PLAIN TEXT BUT HAS THE FORMAT OF AN OBJECT AND WELL KNOWN AS JSON FORMAT AND PANDAS IS USED TO
CREATE A PYTHON FILE TO READ CONTENT OF JSON.
CREATE A CSV FILE (EMPLOYEE.CSV A DATA.JSON)
JSON FILE IS USED TO 
```

```python
import pandas as pd
a=pd.read_json('employee.json')
print(a)                    # AS IN CSV ONLY PRINT FIRST 5 AND LAST 5 DATA.
```

```python
import pandas as pd
a=pd.read_json('employee.json')
print(a.to_string())                         # SAME AS CSV --> TO DISPLAY ALL DATA.
```

```python
#.HEAD:-(UP)
import pandas as pd
a=pd.read_json('employee.json')
print(a.head(10))
```
