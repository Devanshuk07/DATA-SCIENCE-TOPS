```python
'''PANDAS IS USED TO CLEAN CELL.
PANDAS CAN BE USED TO REMOVE ROWS.
PANDAS USE SOME METHOD TO REMOVE ROWS.
DROPNA():- RETURN NEW DATAFRAME AND WILL NOT CHANGE THE ORIGINAL."REMOVES MISSING DATA"
INPLACE():- CHANGE ORIGINAL DATFRAMES DROPNA(INPLACE=TRUE)."NO NEW DATAFRAMES CREATED"
```

```python
import pandas as pd
a=pd.DataFrame({
    'student name' :['DEVANSHU','VIJAY','MAHARAJA','DOOBI','CANDY'],
    'age':[30,20,19,29,'none'],
    'salary':[30000,20000,'none',40000,50000]
})
print(a)
```

```python
#EXAMPLE OF DROPNA():-
import pandas as pd
a=pd.DataFrame({
    'student name' :['DEVANSHU','VIJAY','MAHARAJA','DOOBI','CANDY'],
    'age':[30,20,19,29,None],
    'salary':[30000,20000,None,40000,50000]        # IN NONE N MUST BE CAPITAL.
})
a.dropna(inplace=True)           # WILL NOT TAKE NONE ELEMENT WITH ALL ITS ROW. & T MUST BE CAPITAL IN TRUE.
print(a)
```

```python
# SALARY NONE IS NOT BEEN TAKEN ELSE EVERYTHING NONE IS TAKEN:-
import pandas as pd
a=pd.DataFrame({
    'student name' :['DEVANSHU','VIJAY','MAHARAJA','DOOBI','CANDY'],
    'age':[30,20,19,29,None],
    'salary':[30000,20000,None,40000,50000]
})
a.dropna(subset=['salary'],inplace=True)    #HERE SALARY NONE IS NOT BEEN TAKEN.
print(a)
```

```python
# AGE NONE IS NOT BEEN TAKEN ELSE EVERYTHING NONE IS TAKEN:-
import pandas as pd
a=pd.DataFrame({
    'student name' :['DEVANSHU','VIJAY','MAHARAJA','DOOBI','CANDY'],
    'age':[30,20,19,29,None],
    'salary':[30000,20000,None,40000,50000]
})
a.dropna(subset=['age'],inplace=True)    #HERE AGE NONE IS NOT BEEN TAKEN.
print(a)
```

```python
# FILLNA FILL THE MISSING PLACES WITH 0.0 ITSELF IN PLACE OF NONE.:-
import pandas as pd
a=pd.DataFrame({
    'student name' :['DEVANSHU','VIJAY','MAHARAJA','DOOBI','CANDY'],
    'age':[30,20,19,29,None],
    'salary':[30000,20000,None,40000,50000]
})
a.fillna(0,inplace=True)
print(a)
```

```python
# TAKING MEAN:-
import pandas as pd
a=pd.DataFrame({
    'student name' :['DEVANSHU','VIJAY','MAHARAJA','DOOBI','CANDY'],
    'age':[30,20,19,29,None],
    'salary':[30000,20000,None,40000,50000]
})
a1=a['age'].mean()
a.fillna({'age':a1}, inplace=True) #{'age':a1}:- Only look at the age column, and wherever you find a missing value, replace it with a1
print(a)
```
