```python
'''WHAT IS EXCEPTIONAL HANDELING ?
EXCEPTIONAL HANDELING IS USED TO FIND ENEXCEPTIONAL(LIKE INVALID INPUT OR MISSING FILES) ERROR HANDLING.
BASIC SYNTEX OF EXCEPTION HANDELING:-
TRY:
    STATEMENT
EXCEPT:
    STATEMENT

NOTE:- IT IS NOT LIKE IF CONDITION STATEMENTS.'''
```

```python
'''COMPLETE EXCEPTIONAL HANDELING:-
TRY:-TRY WILL EXECUTE WHEN GIVEN INPUT IS ACCEPTABLE(WHICH DON'T HAVE ANY ERRORS)
EXCEPT:- WILL OCCURE WHEN ERROR COMES.
ELSE:- ALWAYS EXECUTE WHEN TRY HAS NO ERROR.
FINALLY:-ALWAYS EXECUTE.'''
```

```python
# EXAMPLES:-
x = 10
y = 0
print(x / y)   # ZeroDivisionError
print("End")   # This line never runs
```

```python
#EXAMPLE:-(TRY AND ERROR)
try:                      # Code that may cause an error is written inside try.
    x = int(input("Enter a number: "))
    print(10 / x)
except:                   # Runs only if an error occurs inside try. Used to handle the error instead of crashing.
    print("Error occurred")
```

```python
#EXAMPLE:-(TRY AND EXCEPT AND ELSE)
try:
    x = int(input("Enter a number: "))
    print(10 / x)

except ZeroDivisionError:
    print("Division by zero is not allowed")
else:                          # else runs only when try has NO error.else executes only if no exception OCCURS.
    print("Calculation successful")
```

```python
# EXAMPLE:-(TRY EXCEPT ELSE AND FINALLY)
try:
    x = int(input("Enter a number: "))
    print(10 / x)

except ZeroDivisionError:
    print("Division by zero is not allowed")

except ValueError:
    print("Invalid input")

else:                       # ELSE RUNS ONLY WHEN TRY HAS NO ERROR.else executes only if no exception OCCURS.
    print("No errors occurred")

finally:                    #Runs ALWAYS.EVEN ERROR OCCURS OR NOT.
    print("Program finished")
```

```python
try:
    a=int(input("ENTER NUMBER :"))
    b=a/10
    print("YOUR DIVISION RESULT IS :",b)
except:
    print("NOT VALID NUMBER GIVEN")
else:
    print("GOOD REPLY")
finally:
    print("THANK YOU")
```
