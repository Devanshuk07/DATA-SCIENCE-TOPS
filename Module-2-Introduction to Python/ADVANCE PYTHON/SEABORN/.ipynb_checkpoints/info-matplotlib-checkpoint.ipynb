```python
#WHAT IS MATPLOTLIB?(GRAPH IN VISUALITAZTION FORMAT)
'''A MATPLOTLIB IS A LOW LEVEL GRAPH PLOTING LIBRARY IN PYTHON.
-THAT CAN BE USED FOR VISUALIZATION UTILITIES.
-MATPLOTLIB WAS CREATED BY JOHN D. HUNTER.
-MATPLOTLIB IS AN OPEN SOURCE.WE CAN USE IT WITHOUT LISENCE.
-MATPLOTLIB IS MOSTLY WRITTEN IN PYTHON, A FEW SEGMENTS ARE WRITTEN IN C AND JAVA SCRIPT FOR PLATFORM COMPATEBILITY.
```

```python
# what is matplotlib pyplot ?
'''MOST OF ALL MATPLOTLIB UTILITIES LIES UNDER THE PYPLOT as submodule.
-PYPLOT ARE USUALLY IMPORTED UNDER THE PLT ALIAS.
-EXAMPLES:-
```

```python
#draw a line in diagram from position (0,0) to position (6,250).
import matplotlib.pyplot as plt
import numpy as np       # if we want to set any data in figure then numpy is used.
xpoints=np.array([0,6])   # THIS IS HORIZONTAL.
ypoints=np.array([0,250]) # THIS IS VERTICAL.
plt.plot(xpoints,ypoints)
plt.show()
```

```python
#STRAIGHT LINE:-
import matplotlib.pyplot as plt
import numpy as np       # if we want to set any data in figure then numpy is used.
xpoints=np.array([0,0])   # THIS IS HORIZONTAL.
ypoints=np.array([0,25]) # THIS IS VERTICAL.
plt.plot(xpoints,ypoints)
plt.show()
```

```python
# DRAW PLOT WITHOUT LINE:-
import matplotlib.pyplot as plt
import numpy as np       # if we want to set any data in figure then numpy is used.
xpoints=np.array([0,6])   # THIS IS HORIZONTAL.
ypoints=np.array([0,250]) # THIS IS VERTICAL.
plt.plot(xpoints,ypoints,'o')    # small o is used for just rings.no line is drawn.
plt.show()
```
