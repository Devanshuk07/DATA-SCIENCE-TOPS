```python
# OS MODULE IS A MODULE USED IN PYTHON BASICALLY USED TO DELETE A FILE.
# OS MODULE IS USED IN PYTHON BASICALLY FOR EMPTY DIRECTORY REMOVAL.
# OS MODULE IS AN INBUILT MODULE IN PYTHON.
```

```python
# HOW TO CREATE FILE BY OS:-
import os
file=open("regression_model.txt","w")
print("FILE CREATED")
file.close()       # CLOSE IS NECESSARY TO CLOSE THE OPEN FILE SUCCESSFULY OR ELSE YOU CANNOT DELETE IT DUE TO IT IS OPEN.
```

```python
# HOW TO REMOVE FILE BY OS:-
import os
if os.path.exists("regression_model.txt"):          # CHECKS WEATHER PATH FOR PERTICULAR FILE IS THERE OR NOT.
    os.remove("regression_model.txt")               # TO REMOVE THE FILE.
    print("FILE REMOVED")
else:
    print("FILE NOT REMOVED")
```

```python
# REMOVE COMPLETE DIRECTORY(FOLDER):-
import os
if os.rmdir("learning"):
    print("OK REMOVED")
else:
    print("NO....")      # IT WILL DELETE BUT WILL PRINT ELSE ONLY.
```

```python
# HOW TO CREATE A DIRECTORY:-
import os
if os.mkdir("INTRO"):
    print("OK CREATED")
else:
    print("noo...")        ## IT WILL DELETE BUT WILL PRINT ELSE ONLY.
```

```python
# CREATING THE FILE USING FILE HANDELING:-
file=open("INTRO.ipynb","w")
if file:
    print("OK DONE")
else:
    print("NO...")
file.close()
```

```python
# HOW TO WRITE IN A FILE:-
file=open("INTRO.ipynb","w")
b=file.write("'''OS HANDELING IS USED TO MAKE DIRECTORY(FOLDER), DELETE DIRECTORY(FOLDER) AND ALSO REMOVE FILE. IT IS MANDATORY TO WRITE IMPORT OS, NOW TO CREATE =MKDIR(NAME_OF_DIRECTRY) TO REMOVE FILE OR DIRCTRY =RMDIR(NAME_OF_DIRECTORY)")
file.close()
```
