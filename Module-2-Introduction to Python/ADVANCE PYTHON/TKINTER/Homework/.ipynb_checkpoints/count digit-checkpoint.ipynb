```python
a=(input("ENTER THE DIGIT TO COUNT :"))
count=0
for i in a:
    if i.isdigit():    # i is loop variable &&& i.isdigit() is a string method in Python that checks whether a character is a digit (0–9)
        count+=1
print("THE NUMBER OF DIGITS IN INPUT IS : ",count)
```
