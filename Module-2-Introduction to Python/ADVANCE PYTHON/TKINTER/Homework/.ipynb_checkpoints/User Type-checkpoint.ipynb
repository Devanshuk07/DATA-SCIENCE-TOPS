```python
a=input("ENTER YOUR CHOICE")
if a in ["true","false"]:
    print("DATA TYPE : BOOLEAN")
else:
    try:
        int(a)
        print("DATA TYPE : INTEGER")
    except ValueError:
        try:       #it will try to figure wheather the value is float or not
            float(a)
            print("DATA TYPE : FLOAT")
        except ValueError:
                print("DATA TYPE : STRING")    # now every other thing is string so just print no need for try:
```
