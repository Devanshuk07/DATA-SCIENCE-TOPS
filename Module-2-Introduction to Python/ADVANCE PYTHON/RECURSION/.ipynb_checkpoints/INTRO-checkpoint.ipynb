```python
'''RECURSION IS A METHON IN WHICH FUNCTION CALLS ITSELF REPEATEDLY.
EVERY WORK DONE BY RECURSION CAN BE DONE BY LOOPS AND EVERY WORK DONE BY LOOPS CAN BE DONE BY RECURSION.
RECURSION HAS 2ND FUNCTION IN A FUNCTION.
WE CANNOT CALL A FUNCTION IN FUCTION WITHOUT CONDITION,'''
```

```python
#EXAMPLES:-(THIS IS WITHOUT RECURSION)
def show(n):
    print(n)
show(5)
```

```python
#EXAMPLE WITH RECURSION:-
def show(n):
    if n==0:             #IT IS REQUIRED OR ELSE IT WILL PRINT NEGATIVE NUMBER TILL INFINITE.(CONDITION IS MANDATORY OR INFITY VALUE WILL BE PRINTED)
        return           # SO IT WILL BREAK FROM THAT.
    print(n)   #1ST=N=5,,,N=4,,,N=3,,,N=2,,,,N=2,,,N=1
    show(n-1)  # N=5-1=4....TILL N=0------RECURSION HAS 2ND FUNCTION IN A FUNCTION.
show(5)
```

```python
#CALCULATE THE SUM OF FIRST N NUMBERS:-
def sum(n):
    if n==0:
        return 0
    return sum(n-1)+n
sum(10)
```
