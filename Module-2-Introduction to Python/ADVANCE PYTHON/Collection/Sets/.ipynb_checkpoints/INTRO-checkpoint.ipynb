```python
'''WHAT IS SET:- 
SET IS THE COLLECTION OF THE UNORDERED ITEMS.
EACH ELEMENT IN THE SET IS MUTABLE.
IT USES CONSTUCTOR IN PYTHON-SET().
```

```python
set={1,2,3,4,5,"hello","devanshu","devanshu"}
print(set)
print(type(set))
print(len(set))      # no repetaion is taken.
```

```python
a=set()
print(type(a))
```

```python
a=set("hi my name is devanshu")
print(a)     # returns in unordered way.
```

```python
a=set(["devanshu","vijay","rajesh"])
print(a)
```

```python
set=("devanshu","vijay","rajesh")
print(set)
```

```python
a=set("devanshu","vijay","rajesh")
print(a)    #set() does NOT accept multiple arguments.HERE SET WORKS AS FUNCTION AND ONLY RETURN SINGLE VALUE.
```

```python
s = set([1, 2, 3])
print(s)     #this can be given multiple values.IT IS LIST IN SET.
```
