```python
'''DATA COLLECTION OR DATA STRUCTURES IN PYTHON.
DATA STRUCTURES IN PYTHON IS USED TO STORE MULTIPLE DATA.
DATA STORED USING DATA STRUCTURED IN PYTHON ARE SAME TYPE

-string---->""-->MUTABLE
-list(array)--->[](square)-->MUTABLE
-tuple--->()(paranthesis)-->IMMUTABLE
-dictionary--->{}(curly)-->IMMUTABLE
-set--->{}(curly)-->MUTABLE
-array-->{}(curly)-->MUTABLE
'''
```

```python
'''DIFFERNECE BETWEEN DICTIONARY AND SET:-
DICTIONARY USE {} BUT STORE VALUES IN KEY:VALUE FORM.
WHILE SET DON'T HAVE SUCH FORMAT MEANS NO KEY:VALUE REQUIRED''' 
```
