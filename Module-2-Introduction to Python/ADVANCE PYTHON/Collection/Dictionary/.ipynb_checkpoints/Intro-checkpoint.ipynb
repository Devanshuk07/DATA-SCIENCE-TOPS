```python
'''DICTIONARY IS USED TO STORE DATA VALUE IN KEY:VALUE PAIRS.
THEY ARE UNORDERED, MUTABLE(CHANGEABLE) & DON'T ALLOW DUPLICATE KEYS'''
```

```python
#EXAMPLE OF DICTIONARY:-
menu={ "name":"devanshu",    # key is name and value is devanshu. menu is dictionary(dict)
       "marks":90,
       "roll no":7,
       "day":"friday",
       "month":"april",
       12 : 1997,     # we can put int as key
       17.54 : 22,    #we can put float value as key
       "ary":[99,98,97],
     }
menu["surname"]="khakhar"    # you can add the surname here.it means dictionary is mutable(changeable)
menu["name"]="rutvi"         #overwrite the value
print(menu)
print(type(menu))
```
