```python
root={"name":"RUTVI",
     "score":{      #nested dictionary.
    "chem":90,
    "maths":99,
    "sci":98,
     },           # , is nessecary.
      "roll":17,
}
print(root.keys())
print(root.values())     # only values of keys.
print(len(root))         # length means total keys in dict.
print(root.items())      # return all keys:values in tuple(means in parenthithis)
print(root.get("score")) # get value of any key.

root.update({"city":"delhi"})     # to update the dict.{} and in ""
root.update({"country":"INDIA","age":17})      # double update.
print(root)
```

```python
root={"name":"RUTVI",
     "score":{      #nested dictionary.
    "chem":90,
    "maths":99,
    "sci":98,
     },           # , is nessecary.
      "roll":17,
}
for key,value in root.items():
    print(key,value)
```

```python
# MODIFYING THE DICT:-
root={"name":"RUTVI",
     "score":{      #nested dictionary.
    "chem":90,
    "maths":99,
    "sci":98,
     },           # , is nessecary.
      "roll":17,
}
root["name"]="devanshu"              #THIS WILL MODIFY THE DICT.
print(root)
```

```python
# DELETE THE PERTICULAR DICT VALUE:-
root={"name":"RUTVI",
     "score":{      #nested dictionary.
    "chem":90,
    "maths":99,
    "sci":98,
     },           # , is nessecary.
      "roll":17,
}
del root["roll"]               #THIS WILL DELETE THE PERTICULAR KEY WITH ITS VALUE.
print(root)
```

```python
# DELETE THE PERTICULAR DICT VALUE:-(POP DOES SAME AS DELETE)
root={"name":"RUTVI",
     "score":{      #nested dictionary.
    "chem":90,
    "maths":99,
    "sci":98,
     },           # , is nessecary.
      "roll":17,
}
root.pop("roll")               #THIS WILL DELETE THE PERTICULAR KEY WITH ITS VALUE.
print(root)
```

```python
#UPDATING THE DICT:-DONT ALLOW DUPLICATE KEYS/VALUES. MEANS NAME=RUTVI THEN YOU CANNOT GIVE RUTVI,DEVANSHU BOTH THIS IS DUPLICATE.
root={"name":"RUTVI",
     "score":{      #nested dictionary.
    "chem":90,
    "maths":99,
    "sci":98,
     },           # , is nessecary.
      "roll":17,
}
root["name"]="devanshu"               #DUPLICATE KEYS OR VALUES NOT ALLOWED.
print(root)
```
