```python
'''TUPLE IS THE PART OF COLLECTION IN PYTHON.
A TUPLE STORED DATA IN ()--PARENTHISIS.
A TUPLE CAN STORE ALL DATA TYPE.
A TUPLE CAN STORE ANYTYPE OF DATA,EX:-STRING,FLOAT,INT,BOOL...
YOU CANNOT INSERT,UPDATE ,DELETE THE TUPLE bcz as it is IMMUTABLE.'''
```

```python
emp=("devanshu","kumar","amit","rutvi","deep","jay","vijay","ajay")
age=(20,19,29,89,28,27,18,29)
print(emp,"\n",age)
```

```python
emp=["devanshu","kumar","amit","rutvi","deep","jay","vijay","ajay"]
print(tuple(emp))     # to convert list into tuple but type doesn't change.
print(type(emp))
```

```python
#YOU CANNOT ASSIGN VALUE TO TUPLE DIRECTLY:-
emp=("devanshu","kumar","amit","rutvi","deep","jay","vijay","ajay")
age=(20,19,29,89,28,27,18,29)
emp[3]="MEET"
emp[4]="giri"
print(emp,age)
```

```python
#TO UPDATE THE TUPLE YOU FIRST NEED TO CONVERT TUPLE INTO LIST:-
emp=("devanshu","kumar","amit","rutvi","deep","jay","vijay","ajay")
age=(20,19,29,89,28,27,18,29)
a=list(emp)    #TAKES THE TUPLE(EMP) AND CONVERT IT INTO LIST.A IS VARIABLE TO STORE.
a[3]="MEET"
a[4]="giri"
emp=tuple(a)   #TAKING THE UPDATES LIST(EMP)THEN CONVERT LIST BACK TO TUPLE.REASSIGNING THE TUPLE(EMP) TO NEW UPDATED VALUE OF A.

print(a)
```
