```python
# file handeling open file:-
a=open("open.txt","r")
if a:
    print("FILE GET")
else:
    print("ERROR")
```

```python
# file handeling open file:-(GIVING WRONG FILE)
a=open("hello.txt","r")                  # NO SUCH FILE IS THERE IN DIRECTORY/IN FOLDER.
if a:
    print("FILE GET")
else:
    print("ERROR")
```

```python
# file handeling open file:-(R+ --> TO READ AND WRITE BOTH)
a=open("open.txt","r+")
b="hi kd"
if a:
    print("FILE GET",b)    # NOTE--> IT DONT PRINT IN REAL TEXT FILE IT JUST SHOW THERE IS NO ERROR IN OPENING THE FILE(APPEND IS USED TO WRITE IN ORIGINAL FILE)
else:
    print("ERROR")
```
