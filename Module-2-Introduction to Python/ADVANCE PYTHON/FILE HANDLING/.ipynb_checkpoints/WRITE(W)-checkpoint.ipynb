```python
# file handeling open file:-(W-->WRITE ONLY IN EXISTING FILE AND IF FILE NOT FOUND IT WILL CREATE NEW.)
a=open("open.txt","w")
if a:
    print("FILE GET")
else:
    print("ERROR")
```

```python
# file handeling open file:-(W+--> USED TO READ AND WRITE THE EXISTING FILE OR CREATE THE NEW FILE IF NOT FOUND)
a=open("NEW CREATE.txt","w+")     # THE GIVEN FILE NAME DOESN'T EXIST IN THE DIRECTORY. SO IT WILL CREATE NEW.
if a:
    print("FILE CREATED NEW")
else:
    print("ERROR")
```
