```python
'''WHAT IS FILE HANDLIMG IN PYTHON ?
FILE HANDELING IS USED TO HANDLE FILE IN PYTHON.
FILE HANDELING IS USED TO READ,WRITE,CREATE,OPEN AND DELETE FILE USING FUNCTION.
FILE HANDELING METHODS:-
'''
```

```python
'''WHY WE NEED FILE HANDELING?
-STORE USER DATA PERMANENTLY.
-READ CONFIGURATION FILES.
-SAVE LOGS, RESULTS, REPORTS.
-AVOIDS DATA LOSS WHEN PROGRAM STOPS.'''
```

```python
'''TYPES OF FILE IN PYTHON:-
1)TEXT FILES:-
-.txt , .csv , .log
-STORES DATA AS READABLE TEXT.
2)BINARY FILES:-(IMAGES,MUSIC,PDF FILES)
-.jpg , .mp3 , .pdf
-STORE DATA IN BINARY FORMAT.'''
```

```python
'''OPENING A FILE:-
TO OPEN A FILE IN PYTHON-->OPEN() IS USED.
SYNTAX:-
file = open("filename","mode")'''
```

```python
'''HERE MODE:-
MODE MEANS SECOND ARGUMENT FOR FILE OPEN().
-DIFFERENT TYPE OF MODES:-
R    --->   READ
W    --->   WRITE
A    --->   APPEND
X    --->   CREATE A NEW FILE.
B    --->   BINARY MODE(IMAGES,MUSIC,PDF FILES).
T    --->   TEXT.
NOTE :- ONLY USED AFTER OPEN()
FILE=OPEN("FILE_NAME","MODE")'''
```

```python
'''R    --->   READS
SYNTAX:-
file=open("file_name","r")   --->OPEN FILE("FILE NAME", "MODE")
content=file.read()     ---> READS ENTIRE CONTENT OF THE FILE.  CONTENT==VARIABLE
print(content)          ---> PRINTS FILE DATA ON SCREEN.
file.close()            --->CLOSE THE FILE. VERY IMP TO FREE SYSTEM MEMORY

READLINE ALSO USED :-
SYNTAX:-
file = open("file_name", "r")
print(file.readline())     --->READ SINGLE LINE FROM THE FILE.
file.close()

READLINES ALSO USED:-
SYNTAX:-
file = open("data.txt", "r")
lines = file.readlines()     --->READ ENTIRE FILE.EACH LINE BECOME ONE ELEMENT IN A LIST.
print(lines)
file.close()
'''
```

```python
'''W     --->   WRITE
SYNTAX:-
file = open("file_name", "w")
file.write("Hello Python")
file.close()

NOTE:- IF OLD DATA IN DATA.TXT EXITS IT WILL BE DELETED AND NEW DATA(GIVEN BY YOU) WILL BE STORED.
```

```python
'''A      --->  APPEND
SYNTAX:-
file = open("data.txt", "a")
file.write("\nThis line is added")
file.close()

NOTE:- WRITE DELETE OLD DATA AND ADD NEW ONE. BUT APPEND KEEP OLD DATA AND ADD NEW AT THE END.
```

```python
'''X   --->   CREATE NEW FILE
SYNTAX:-
file = open("file_name", "x")        ---> "X"-CREATE A FILE
file.write("This file is newly created")  --> WRITE IN FILE.
file.close()

NOTE:-FILE IS CREATED AND DATA IS WRITTEN.
```

```python
'''B      --->   BINARY( MEANS IMAGE , MUSIC , PDF FILES)
BINARY USE 4 MODES IN IT:-
1)rb    --->   READ BINARY
2)wb    --->   WRITE BINARY
3)ab    --->   APPEND BINARY
4)xb    --->   CREATE BINARY

1)rb    --->   READ BINARY:-
SYNTAX:-(TWO WAYS)

(IN THIS CODE, FILE AUTOMATICALLY CLOSED,SAFE EVEN IF ERROR OCCURES)

with open("photo.jpg", "rb") as file:    --> WITH AUTOMATTICALLY CLOSE THE FILE SAFELY AFTER WRITTING.
    data = file.read()
    print(data)

    OR
    
(YOU HAVE TO MANUAALY CLOSE THE FILE IN THIS CODE GIVEN BELOW)

file = open("photo.jpg", "rb")
data = file.read()       ---> READS ENTIRE IMAGE FILE.
file.close()

2)wb   --->  WRITE BINARY:-
SYNTAX():-

with open("photo.jpg", "rb") as src:
    data = src.read()

with open("copy.jpg", "wb") as dest:
    dest.write(data)          ---> WRITE FROM PHOTO.JPG --> INTO ---> COPY.JPG.

    OR

with open("photo.jpg", "rb") as source:
    data = source.read()

with open("copy.jpg", "wb") as target:
    target.write(data)

BOTH SAME JUST DIFFERNECE IS dest ---> target WORKS SAME.

3)ab   --->   APPEND BINARY:-
SYNTAX:-

```
