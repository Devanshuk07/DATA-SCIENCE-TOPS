```python
import datetime as datetime  
a=datetime.date.today()       # syntax for only date
print(a)
```

```python
import datetime as datetime   #for time only as datetime is mandatory
a=datetime.datetime.now().time() # syntax for only time
print(a)
```

```python
import datetime as datetime
a=datetime.datetime.now().strftime("%M:%H:%S")   # for  formation of time means hour:min:sec  STRFTIME IS METHOD.
print(a)
```

```python
import datetime as datetime
a=datetime.datetime.now().strftime("%d-%m-%Y")   # for  formation of date but in small letters only. big Y=full year,,,d and m always small
print(a)
```

```python
import datetime as datetime
a=datetime.datetime.now().strftime("%A")   # A for full name of day small a for tue short name
print(a)
```
