```python
root={"name":"RUTVI",
     "score":{      #nested dictionary.
    "chem":90,
    "maths":99,
    "sci":98,
     },           # , is nessecary.
      "roll":17,
}
print(root.keys())
print(root.values())     # only values of keys.
print(len(root))         # length means total keys in dict.
print(root.items())      # return all keys:values in tuple(means in parenthithis)
print(root.get("score")) # get value of any key.

root.update({"city":"delhi"})     # to update the dict.{} and in ""
root.update({"country":"INDIA","age":17})      # double update.
print(root)
```
