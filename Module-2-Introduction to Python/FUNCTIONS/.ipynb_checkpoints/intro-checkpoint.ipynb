```python
def calc(a,b):      # only take variables dont assign values to it.
    sum=a+b
    print(sum)      # but this will not print the sum.In output.BUT NECCESARY WHEN YOU CALL THE FUNTION IT WILL PRINT.

calc(2,10)          #here call the function to print and then give (values in ()of function)
calc(12,20)
calc(80,100)        # like this no need to assign differently just call funtion give value and then it will sum of it.
```
