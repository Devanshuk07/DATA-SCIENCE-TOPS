```python
# nested function
#function within function is called nested function
# SYNTAX OF NESTED FUNCTION

def outer(a,b):       # function
    c=a+b
    print(c)
    def inner(a,b):   #function inner function
        c=a-b
        print(c)
    inner(10,20)     # always in line of def
outer(20,90)
```
