```python
# EXAMPLES :-
collection=set()      # set() represent the sets and not dictionary
#collection={}        means {}--it will take dictionary.
collection.add(1)
collection.add(2)
collection.add(3)
collection.add("Devanshu")   # add--string
collection.add((9,8,7,6,5))  #add--tuple.

print(collection)
```

```python
collection=set()      # set() represent the sets and not dictionary.
collection.add(1)
collection.add(2)
collection.add(3)

collection.remove(2)    # removes element.
print(collection)
```

```python
collection=set()
collection.add(1)
collection.add(2)
collection.add(3)
collection.add("Devanshu")   
collection.add((9,8,7,6,5))  
collection.clear()          #clear the all set.
print(collection)
```
