```python
'''For a Pandas Data Frame created from a NumPy array, what is the default behavior for 
the labels for the columns? For the rows?
```

```python
import pandas as pd
a=pd.DataFrame([[10,20,30],
               [40,50,70]
               ])
print(a)
```

```python
'''SO BY DEFAULT THE COLUMNS AND ROW NAMES ARE SEQUENTIAL INTEGERS...unless YOU PROVIDE NAME TO IT.'''
```

```python
#GIVING NAMES TO COLUMNS and ROWS:-
import pandas as pd
a=pd.DataFrame([[10,20,30],
               [40,50,70]
               ])
a.columns=['col1','col2','col3']
a.index=['row1','row2']
print(a)
```
