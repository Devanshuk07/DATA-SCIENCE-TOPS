```python
# Using NumPy, create a Pandas Data Frame with five rows and three columns.
```

```python
import pandas as pd
import numpy as np
a=np.array([[10,20,30],           # pd.array() is meant for 1D arrays only, not for 2D-multiple rows and columns.
           [40,50,60],            # to take 2d or 3d array we have to import numpy as np and then np.array
           [70,80,90],
            [18,29,39],
            [49,59,69]])
b=pd.DataFrame(a, columns=['Col1','Col2','Col3'])
print(b)
```

```python
# IF YOU TAKE IMPORT PANDAS THEN DONT USE ARRAY USE SERIES..
import pandas as pd

a = pd.Series([10, 20, 30, 40])
print(a)
```
