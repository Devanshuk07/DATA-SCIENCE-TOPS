```python
# Compute the minimum, 25th percentile, median, 75th, and maximum of ser.
```

```python
import pandas as pd
a=pd.Series([11,22,33,44,55,66,77,88,99])

print("MINIMUM:",a.min())
print("25th PERCENTILE:",a.quantile(0.25))
print("75th PERCENTILE:",a.quantile(0.75))
print("MEDIAN:",a.median())
print("MAXIMUM:",a.max())
```
