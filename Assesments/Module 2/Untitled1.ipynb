```python
# Hotel Menu Program

menu = {
    "Tea": 20,
    "Coffee": 30,
    "Sandwich": 50,
    "Burger": 80,
    "Pizza": 120
}

print("🍽️ Welcome to Python Hotel 🍽️")
print("\n------ MENU ------")
for item, price in menu.items():
    print(f"{item}: ₹{price}")
print("------------------")

total = 0

while True:
    choice = input("\nEnter item name to order (or type 'done' to finish): ").title()
    
    if choice == "Done":
        break
    elif choice in menu:
        qty = int(input(f"Enter quantity of {choice}: "))
        cost = menu[choice] * qty
        total += cost
        print(f"Added {qty} {choice}(s) - ₹{cost}")
    else:
        print("❌ Item not available. Please choose from the menu.")

print("\n🧾 Final Bill")
print("Total Amount to Pay: ₹", total)
print("🙏 Thank you! Visit again.")
```
