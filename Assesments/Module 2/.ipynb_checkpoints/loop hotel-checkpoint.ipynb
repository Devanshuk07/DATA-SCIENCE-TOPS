```python
# In dictionaries, every key has a value associated with it.
import sys
menu={
    "PANEER TIKKA MASALA":350,
     "PANEER BHURJEE":290,
     "MASALA DHOSA":110,
    "CHEESE CHILLI MASALA DHOSA":150,
     "CHINESE BHEL":90,
     "MANCHURIYAN DRY":110,
     "GUJARATI THALI FIXED":250,
     "UTTAPAM":100,
     "PANI PURI":30,
     "SHAHI AALOO":80
}
for item, price in menu.items():     # menu.items is in dictionary so it is mandatory.
    #key is taken as variable and it represent names. price is taken as variable means marks(you can take anything there)
    print(item,":",price)    # to print two thing together , is used in betwenn
total=0  # start the total bill=0;

while True: # while need a condition or True Or False(T and F capital)or else it will give syntax error
    choice=input("ENTER ITEM NAME (DONE TO ORDER/CANCEL TO CANCEL ANYTIME) :").upper()  #.upper() helps to convert user input to upper case.
    
    if choice=="CANCEL":
        print("YOUR ORDER HAS BEEN CANCELLED......")
        sys.exit()  # stops the whole program
    if choice=="DONE":
        break
    if choice in menu:
        qty=int(input("ENTER THE QUANTITY YOU WANT TO EAT :"))
        cost=menu[choice]*qty   # menu[choice] means if user choice masala dhosa it will take the price of it like 110 and then as per user provide qty it will *it with it.
        # every key has a value associated with it.
        #So menu[choice] always gives the value(price), never the key(name) itself.
        total=total+cost
        print("ITEM ADDED SUCCESSFULLY 😍")
    else:
        print("ITEM UNAVIALABLE😒")
    
print("\nYOUR TOTAL IS :",total)
print("\nTHANK YOU DO VISIT GAIN")
```
