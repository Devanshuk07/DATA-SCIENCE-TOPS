```python
a=str(input("ENTER THE STRING :"))
b=str(input("ENTER THE SUBSTRING :"))
print(a+" "+b)
if len(a)<2 and len(b)<2:
    print("ERROR !!LENGTH MUST BE MORE THAN TWO CHARCTORS")
else:
    new=a[:2]+b[:2]    # new=a[:2] means two charctor from string a and + 2 charactor from str b.
    print("RESULT =",new)
```
